```python
import xarray as xr
from matplotlib import pyplot as plt
import cartopy.crs as ccrs
import numpy as np
```

```python
dem_path = '/work/FAC/FGSE/IDYST/tbeucler/downscaling/fquareng/data/OPERA/DEM/europe_1km_dem.nc'

dem_ds = xr.open_dataset(dem_path)
dem_ds.attrs

def plot_dem_netcdf(dem_path):
    ds = xr.open_dataset(dem_path)
    dem = ds["DEM"]

    plt.figure(figsize=(10, 8))
    plt.imshow(dem[::-1, :], origin='upper', cmap='terrain')

    plt.colorbar(label='Elevation (m)')
    plt.title('Digital Elevation Model')
    plt.xlabel('Longitude')
    plt.ylabel('Latitude')

    return dem

o_dem = plot_dem_netcdf(dem_path)
```

```python
dem_path = '/work/FAC/FGSE/IDYST/tbeucler/downscaling/fquareng/data/OPERA/DEM/europe_1km_dem.nc'
dem_ds = xr.open_dataset(dem_path)

dem = dem_ds["DEM"]
dem_lat = dem_ds["y"]
dem_lon = dem_ds["x"]

x = np.where(dem<0)[1]
y =  np.where(dem<0)[0]

# x = np.where(dem==np.nanmin(dem))[1]
# y = np.where(dem==np.nanmin(dem))[0]

_, ax = plt.subplots(figsize=(15, 15), subplot_kw={'projection': ccrs.Mercator()})
im_dem = ax.pcolormesh(dem_lon, dem_lat, dem, transform=ccrs.Mercator(), cmap='terrain', alpha=0.5)
ax.scatter(dem_lon[x], dem_lat[y], c="r")
```
